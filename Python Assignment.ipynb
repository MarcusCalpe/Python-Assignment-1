# Problem 1: Word Rotation Problem

```python
def rotate_word(text):
    result = ""
    for i in range(len(text)):
        result = text[i] + result
    return result

print(rotate_word("Test"))
```

# Problem 2: Username Builder Problem

```python
def make_username(f,l):
    first_name = f.lower().replace(" ","")
    last_name = l.lower().replace(" ","")
    result = first_name + "." + last_name
    return result

make_username("Marcus Nathan", "Calpe")
```

# Problem 3: Bookend Swap Problem

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]

swap_bookends([1, 2, 3, 4, 5, 6])
```
